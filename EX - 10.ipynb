```python
import math
import re
import pickle
from collections import Counter

# DOMAIN-SPECIFIC DATASET

texts = [
    'The movie was fantastic and inspiring',
    'Excellent acting and wonderful story',
    'I really enjoyed the film',
    'The movie was boring and too long',
    'Terrible screenplay and poor acting',
    'I did not like the movie at all',
    'Amazing direction and great music',
    'The plot was disappointing and dull'
]

labels = [1, 1, 1, 0, 0, 0, 1, 0]

# TRAIN-TEST SPLIT

X_train = texts[:6]
y_train = labels[:6]

X_test = texts[6:]
y_test = labels[6:]

# TEXT VECTORIZATION

def tokenize(text):
    return re.findall(r'\b\w+\b', text.lower())

word_counts = Counter()

for text in X_train:
    word_counts.update(set(tokenize(text)))

vocabulary = sorted(word_counts.keys())

def vectorize(text):
    words = set(tokenize(text))
    return [1 if word in words else 0 for word in vocabulary]

X_train_vec = [vectorize(text) for text in X_train]
X_test_vec = [vectorize(text) for text in X_test]

# SIMPLE SENTIMENT MODEL

positive_words = {
    'fantastic', 'inspiring', 'excellent', 'wonderful',
    'enjoyed', 'amazing', 'great'
}

negative_words = {
    'boring', 'terrible', 'poor', 'disappointing',
    'dull', 'not', 'like'
}

def predict_sentiment(text):
    words = tokenize(text)

    positive_score = sum(word in positive_words for word in words)
    negative_score = sum(word in negative_words for word in words)

    if positive_score >= negative_score:
        return 1
    else:
        return 0

# PREDICTION

y_pred = [predict_sentiment(text) for text in X_test]

# EVALUATION

correct = sum(
    actual == predicted
    for actual, predicted in zip(y_test, y_pred)
)

accuracy = correct / len(y_test)

print('===== MODEL EVALUATION =====')
print('Accuracy:', round(accuracy, 3))

print('\nActual:', y_test)
print('Predicted:', y_pred)

# SAVE MODEL

model_data = {
    'positive_words': positive_words,
    'negative_words': negative_words,
    'vocabulary': vocabulary
}

with open('sentiment_model.pkl', 'wb') as file:
    pickle.dump(model_data, file)

print('\nModel saved as sentiment_model.pkl')

# TEST ON NEW SENTENCE

new_text = 'The film was absolutely wonderful'

prediction = predict_sentiment(new_text)

label_name = 'Positive' if prediction == 1 else 'Negative'

print('\n===== NEW TEXT PREDICTION =====')
print('Text:', new_text)
print('Predicted Sentiment:', label_name)
```

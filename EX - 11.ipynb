```python
# Ex. No. 11
# AI-BASED CONTENT GENERATION SYSTEM FOR TEXT, IMAGE AND MULTIMEDIA APPLICATIONS
# Educational Working Version

from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

# --------------------------------------------------
# USER TOPIC
# --------------------------------------------------

topic = 'The benefits of renewable energy'

# --------------------------------------------------
# 1. TEXT GENERATION
# --------------------------------------------------

generated_text = (
    'Renewable energy sources such as solar and wind power reduce carbon emissions, '
    'improve environmental sustainability, and help create a cleaner future for '
    'coming generations.'
)

print('===== GENERATED TEXT =====')
print(generated_text)

# Save text

with open('content_text.txt', 'w') as f:
    f.write(generated_text)

print('\nText saved as content_text.txt')

# --------------------------------------------------
# 2. IMAGE GENERATION
# --------------------------------------------------

img = Image.new(
    'RGB',
    (512, 512),
    color=(220, 255, 220)
)

draw = ImageDraw.Draw(img)

draw.text(
    (20, 20),
    'AI Generated Content',
    fill='black'
)

draw.text(
    (20, 70),
    topic,
    fill='black'
)

draw.text(
    (20, 130),
    'Solar Energy',
    fill='green'
)

draw.text(
    (20, 180),
    'Wind Energy',
    fill='blue'
)

draw.text(
    (20, 230),
    'Clean Environment',
    fill='black'
)

img.save('content_image.png')

print('Image saved as content_image.png')

# Display image

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.axis('off')
plt.title('AI Generated Image')
plt.show()

# --------------------------------------------------
# 3. MULTIMEDIA CONTENT
# --------------------------------------------------

print('\n===== MULTIMEDIA CONTENT =====')

multimedia_description = (
    'Renewable energy can be presented using generated text, '
    'visual content, and audio narration.'
)

print(multimedia_description)

# Save multimedia description

with open('multimedia_description.txt', 'w') as f:
    f.write(multimedia_description)

print('\nMultimedia description saved as multimedia_description.txt')

# --------------------------------------------------
# FINAL OUTPUT FILES
# --------------------------------------------------

print('\n===== GENERATED FILES =====')

print('1. content_text.txt')
print('2. content_image.png')
print('3. multimedia_description.txt')
```
